# Exercício — Classificação de Flores com KNN

## Objetivo

Neste exemplo, você irá utilizar o algoritmo **K-Nearest Neighbors (KNN)** para classificar flores em três espécies:

- *Iris setosa*
- *Iris versicolor*
- *Iris virginica*

O dataset possui quatro atributos:

- comprimento da sépala;
- largura da sépala;
- comprimento da pétala;
- largura da pétala.

A atividade foi organizada para que você possa se concentrar no funcionamento do **KNN**, sem precisar realizar etapas de pré-processamento dos dados.



```python

import numpy as np

def distancia_euclidiana(a, b):
    return np.sqrt(np.sum((a - b) ** 2))
```

```python
def knn_predict(X_train, y_train, X_test, k=3):

    previsoes = []

    for x in X_test:

        distancias = []

        for i in range(len(X_train)):
            distancia = distancia_euclidiana(x, X_train[i])
            distancias.append((distancia, y_train[i]))

        distancias.sort(key=lambda x: x[0])

        vizinhos = distancias[:k]

        classes = [classe for distancia, classe in vizinhos]

        classes_unicas, contagens = np.unique(
            classes,
            return_counts=True
        )

        classe_predita = classes_unicas[np.argmax(contagens)]

        previsoes.append(classe_predita)

    return np.array(previsoes)
```

## Carregando o dataset

```python
from sklearn.datasets import load_iris

iris = load_iris()

X = iris.data
y = iris.target
```

## Separando treinamento e teste

Agora vamos dividir os dados em:

- **70% para treinamento**
- **30% para teste**

A opção `stratify=y` garante que a proporção das classes seja aproximadamente mantida nos dois conjuntos.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("Treinamento:", X_train.shape[0], "instâncias")
print("Teste:", X_test.shape[0], "instâncias")
```

## Treinando o KNN

Vamos começar utilizando **K = 3**.

O KNN irá classificar cada nova instância observando os seus três vizinhos mais próximos.

```python
k = 3

y_pred = knn_predict(
    X_train,
    y_train,
    X_test,
    k=k
)

print("Previsões:")
print(y_pred)

print("Labels reais:")
print(y_test)
```

## Calculando a taxa de acertos (Acurácia)

```python
acertos = np.sum(y_pred == y_test)

acuracia = acertos / len(y_test)

print(f"Acurácia: {acuracia:.3f}")
```

## Exemplo com o scikit learn

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3)

knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)

print("Previsões:", y_pred)
```

## Avaliando o classificador

Vamos calcular a acurácia do modelo.

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print(f"Acurácia: {accuracy:.3f}")
```

## Matriz de confusão

Agora vamos analisar quais classes foram confundidas pelo modelo.

```python
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix
import matplotlib.pyplot as plt


cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=iris.target_names
)

disp.plot()
plt.title("Matriz de Confusão — KNN")
plt.show()
```
