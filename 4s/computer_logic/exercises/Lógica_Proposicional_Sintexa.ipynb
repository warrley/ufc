## 1. Relembrando a gramática

A gramática nos diz como podemos construir uma fórmula.

$
\varphi ::= P
\mid
(\neg\varphi)
\mid
(\varphi\land\varphi)
\mid
(\varphi\lor\varphi)
\mid
(\varphi\rightarrow\varphi)
$


Uma fórmula pode ser construída de cinco formas:

1. uma proposição atômica;
2. uma negação de uma fórmula;
3. uma conjunção de duas fórmulas;
4. uma disjunção de duas fórmulas;
5. uma implicação de duas fórmulas.

### Exemplos

São fórmulas corretas:

$
P
$

$
(\neg P)
$

$
(P\land Q)
$

$
(P\lor Q)
$

$
(P\rightarrow Q)
$

$
((P\rightarrow Q)\rightarrow R)
$

A última fórmula é construída recursivamente: primeiro construímos fórmulas menores e depois usamos essas fórmulas para construir uma fórmula maior.

## 2. Como transformar a gramática em Python?

Vamos começar com uma ideia muito simples.

Na gramática temos:

$
\varphi ::= P
$

Isso significa:

> Uma fórmula pode ser uma proposição atômica.

Em Python, podemos representar isso simplesmente como:

```python
return "P"

Vamos criar uma função chamada formula():

def formula():
    return "P"

````python

---

```python
def formula():
    return "P"

print(formula())
````

### O que aconteceu?

Nossa função `formula()` representa uma parte da gramática:

$
\varphi ::= P
$

Quando chamamos formula(), ela retorna $P$

Mas nossa gramática possui outras possibilidades.

Precisamos ensinar o programa a construir também:

$(\neg \varphi)$

$(\varphi \wedge \varphi)$

$(\varphi \vee \varphi)$

$(\varphi \rightarrow \varphi$)


É aqui que entra a **recursividade**.


## 3. Uma função que chama a própria função

def formula():

## 4. Vamos começar pela negação

Uma das regras da gramática é:

$\varphi ::= (\neg\varphi)
$

Podemos traduzir essa regra para Python:

```python
def formula():
    parte = formula()
    return "(¬" + parte + ")"```



Mas existe um problema!

Se executarmos essa função, ela nunca vai parar.

Precisamos dizer ao programa **quando parar a recursão**.

Esse ponto é chamado de **caso base**.



## 5. Caso base

Na gramática, temos uma regra que não precisa de outra fórmula:

$
\varphi ::= P
$

Essa será nossa **parada**.

Para facilitar, vamos pedir à função um número chamado `nivel`.

- Se `nivel == 0`, retornamos `P`.
- Se `nivel > 0`, podemos construir uma fórmula maior.

````python
#Implementando BNF - Lógica Proposicional
#átomo e negação
def formula(nivel):
    if nivel == 0:
        return "P"
    else:
        return "(¬" + formula(nivel - 1) + ")"

print(formula(0))
print(formula(1))
print(formula(2))
print(formula(3))
````

````python
#Implementando BNF - Lógica Proposicional
#átomo, negação e implicação
def formula(nivel):
    if nivel == 0:
        return "P"
    else:
        esquerda = formula(nivel - 1)
        direita = formula(nivel - 1)
        return "(" + esquerda + " → " + direita + ")"

print(formula(0))
print(formula(1))
print(formula(2))
````

````python
#Implementando BNF - Lógica Proposicional
#Ampliando o número de símbolos e conectivos


import random

proposicoes = ["P", "Q"]

def formula(nivel):

    # Caso base
    if nivel == 0:
        return random.choice(proposicoes)

    # Escolha uma regra da gramática
    regra = random.choice(["negacao", "conjuncao", "implicacao"])

    if regra == "negacao":
        parte = formula(nivel - 1)
        return "(¬" + parte + ")"

    elif regra == "conjuncao":
        esquerda = formula(nivel - 1)
        direita = formula(nivel - 1)
        return "(" + esquerda + " ∧ " + direita + ")"

    else: #implicação
        esquerda = formula(nivel - 1)
        direita = formula(nivel - 1)
        return "(" + esquerda + " → " + direita + ")"


for i in range(10):
    print(formula(3))
````
